```python
import numpy as np
from scipy import signal
```

```python
with open("input.txt") as f:
  inp = f.read()
```

```python
a = np.array([[1  if el == "@" else 0 for el in r] for r in inp.split()])
a.shape
```

```python
ker = np.array([[1,1,1],[1,0,1],[1,1,1]])
```

```python
c = signal.convolve2d(a, ker, mode='same')
```

```python
s = ((a == 1) & (c < 4)).sum()
```

```python
inter = a.copy()
res = 0
```

```python
while True:
  c = signal.convolve2d(inter, ker, mode='same')
  mask = (inter == 1) & (c < 4)
  s = mask.sum()
  if s == 0: break
  res += s
  inter[mask] = 0

res
```
