## part 1
how many times does the dial end up at 0

```python
s = 50
mod = 100
counter = 0
for l in inp.split():
  d = l[0]
  n = int(l[1:])
  # print(d,n)
  if d == 'R':
    s += n
  else:
    s -= n
  if s % mod == 0:
    counter += 1
counter
```

## part 2
how many times does the dial cross 0 (stopping at 0 also counts)
let's consider a few scenarios:
- turning right we cross 0 if `s` + `n` >= mod. in fact, the number of times we cross is exactly the floor division by mod.
    - 101 // 100 = 1, one time.
- turning left we cross 0 if `s` - `n` <= 0. We can turn it into the same eq by switching sides, we get `n` -`s` >= 0, and using modular stuff we can make this `n` + (-s % mod) >= mod

okay, let's test that second thing to make sure im not wrong

```python
ts = 5
tn = 40
tmod = 100
ts - tn, tn + (- ts % mod), (tn + (- ts % mod)) // tmod
```

```python
def cc(s, n, mod):
  assert 0 <= s < mod
  assert mod > 0
  if n < 0:
    n = -n
    s = - s % mod
  return (n + s) // mod
cc(ts, -tn, mod)
```

```python
assert cc(1, 9, 10) == 1
assert cc(2, -3, 10) == 1
assert cc(5, 20, 10) == 2
assert cc(50, 1000, 100) == 10
```

```python
s = 50
mod = 100
counter = 0
print(s,  counter)
for l in inp.split():
  d = l[0]
  n = int(l[1:])
  sgn = 1 if d == 'R' else -1
  counter += cc(s % mod, n * sgn, mod)
  
  s = (s + sgn * n) % mod
  # print(s, l, counter)
counter
```
